```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# 1. Load the dataset
df = pd.read_csv('diabetes.csv')

# 2. Data Cleaning: In this dataset, 0 in certain columns represents missing data
zero_features = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for feature in zero_features:
    df[feature] = df[feature].replace(0, df[feature].median())

# 3. Separate features (Drop the target 'Outcome' for unsupervised learning)
X = df.drop(columns=['Outcome'])

# 4. Feature Scaling (Crucial for distance-based clustering algorithms)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Data preprocessed successfully. Shape: {X_scaled.shape}")
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Initialize PCA to retain 2 components for visualization
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Check explained variance
variance_retained = pca.explained_variance_ratio_
print(f"Variance explained by PC1: {variance_retained[0]:.2%}")
print(f"Variance explained by PC2: {variance_retained[1]:.2%}")

# Create a DataFrame for easy plotting
pca_df = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Find optimal clusters using Silhouette Score
for k in range(2, 6):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(f"K-Means Silhouette Score for K={k}: {score:.4f}")

# Fit final model (Choosing K=2 for demo)
kmeans_final = KMeans(n_clusters=2, init='k-means++', random_state=42)
pca_df['KMeans_Labels'] = kmeans_final.fit_predict(X_scaled)

# Plot K-Means
plt.figure(figsize=(8, 5))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['KMeans_Labels'], cmap='viridis', s=40, alpha=0.8)
plt.title('K-Means Clustering (Visualized via PCA)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

# 1. Generate the Tree Structure (Dendrogram)
plt.figure(figsize=(12, 6))
plt.title('Hierarchical Clustering Tree Structure (Dendrogram)')
plt.xlabel('Samples Index')
plt.ylabel('Euclidean Distances')

# Using 'ward' linkage to minimize variance within clusters
dendrogram = sch.dendrogram(sch.linkage(X_scaled, method='ward'), truncate_mode='lastp', p=30) 
plt.show()

# 2. Fit Agglomerative Model
hc = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
pca_df['HC_Labels'] = hc.fit_predict(X_scaled)

# Plot Hierarchical Clusters
plt.figure(figsize=(8, 5))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['HC_Labels'], cmap='plasma', s=40, alpha=0.8)
plt.title('Hierarchical Clustering (Visualized via PCA)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
from sklearn.cluster import DBSCAN

# eps: maximum distance between two samples to be considered neighbors
# min_samples: number of samples in a neighborhood for a point to be a core point
dbscan = DBSCAN(eps=1.5, min_samples=5)
pca_df['DBSCAN_Labels'] = dbscan.fit_predict(X_scaled)

# Check number of clusters formed (ignoring noise code -1)
n_clusters = len(set(pca_df['DBSCAN_Labels'])) - (1 if -1 in pca_df['DBSCAN_Labels'] else 0)
n_noise = list(pca_df['DBSCAN_Labels']).count(-1)
print(f"DBSCAN estimated clusters: {n_clusters}")
print(f"DBSCAN estimated noise points: {n_noise}")

# Plot DBSCAN
plt.figure(figsize=(8, 5))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['DBSCAN_Labels'], cmap='tab10', s=40, alpha=0.8)
plt.title('DBSCAN Clustering (Noise labeled as -1)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster Tag')
plt.show()
```

```python
from sklearn.cluster import MeanShift, estimate_bandwidth

# Estimate the bandwidth parameter automatically
bandwidth = estimate_bandwidth(X_scaled, quantile=0.2, n_samples=500)

# Fit Mean Shift
meanshift = MeanShift(bandwidth=bandwidth, bin_seeding=True)
pca_df['MeanShift_Labels'] = meanshift.fit_predict(X_scaled)

print(f"Mean Shift estimated clusters: {len(np.unique(meanshift.labels_))}")

# Plot Mean Shift
plt.figure(figsize=(8, 5))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['MeanShift_Labels'], cmap='rainbow', s=40, alpha=0.8)
plt.title('Mean Shift Clustering (Visualized via PCA)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.mixture import GaussianMixture

# 1. Initialize Gaussian Mixture Model
# We select n_components=2 to align with the typical 2-class nature of the dataset
gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)

# 2. Fit the model and predict cluster labels
pca_df['GMM_Labels'] = gmm.fit_predict(X_scaled)

# 3. Get the probabilities of cluster assignments (Soft Clustering)
probs = gmm.predict_proba(X_scaled)
print("Sample probabilities for the first 5 rows:\n", probs[:5])

# 4. Plot GMM Clustering Results (Visualized via PCA components)
plt.figure(figsize=(8, 5))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['GMM_Labels'], cmap='coolwarm', s=40, alpha=0.8)
plt.title('Gaussian Mixture Model (GMM) Clustering')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Probability/Cluster Assignment')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score

# 1. Plotting the Elbow Curve
inertia = []
K_range = range(1, 11)
for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))

ax[0].plot(K_range, inertia, marker='o', linestyle='--', color='b')
ax[0].set_title('The Elbow Method showing the Optimal K')
ax[0].set_xlabel('Number of Clusters (K)')
ax[0].set_ylabel('Inertia (Within-cluster Sum of Squares)')
ax[0].grid(True)

# 2. Plotting the Silhouette Coefficients for K=2
k_chosen = 2
km_final = KMeans(n_clusters=k_chosen, init='k-means++', random_state=42)
cluster_labels = km_final.fit_predict(X_scaled)
silhouette_avg = silhouette_score(X_scaled, cluster_labels)
sample_silhouette_values = silhouette_samples(X_scaled, cluster_labels)

y_lower = 10
for i in range(k_chosen):
    ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]
    ith_cluster_silhouette_values.sort()
    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    color = plt.cm.nipy_spectral(float(i) / k_chosen)
    ax[1].fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_silhouette_values,
                        facecolor=color, edgecolor=color, alpha=0.7)
    ax[1].text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
    y_lower = y_upper + 10

ax[1].set_title(f'Silhouette Plot for K={k_chosen} (Avg Score: {silhouette_avg:.2f})')
ax[1].set_xlabel('Silhouette Coefficient Values')
ax[1].set_ylabel('Cluster Label')
ax[1].axvline(x=silhouette_avg, color="red", linestyle="--")
ax[1].set_yticks([]) 

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Scikit-Learn and SciPy imports
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN, MeanShift, estimate_bandwidth
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
import scipy.cluster.hierarchy as sch

print("--- Step 1: Loading & Preprocessing Data ---")

# Check if file exists to prevent FileNotFoundError
file_name = 'diabetes.csv'
if not os.path.exists(file_name):
    raise FileNotFoundError(f"⚠️ '{file_name}' file aapke current folder mein nahi mila! "
                            f"Kindly check karein ke aapki Python script aur CSV file aik hi folder mein hain.")

df = pd.read_csv(file_name)

# Handling zeros in clinical features (replacing with median values)
zero_features = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for feature in zero_features:
    df[feature] = df[feature].replace(0, df[feature].median())

# Separate features (Dropping the 'Outcome' target label for unsupervised learning)
X = df.drop(columns=['Outcome'])

# Feature Scaling (Standardization)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


print("--- Step 2: Dimensionality Reduction (PCA & t-SNE) ---")
# 1. PCA for linear projection
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# 2. t-SNE for non-linear projection (FIXED: Removed n_iter to support all scikit-learn versions)
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

# Master dataframe to hold original components and all predictions
summary_df = pd.DataFrame({
    'PC1': X_pca[:, 0],
    'PC2': X_pca[:, 1],
    'tSNE1': X_tsne[:, 0],
    'tSNE2': X_tsne[:, 1]
})


print("--- Step 3: Training All Unsupervised Models ---")

# 1. K-Means
kmeans = KMeans(n_clusters=2, init='k-means++', random_state=42)
summary_df['KMeans'] = kmeans.fit_predict(X_scaled)

# 2. Hierarchical Clustering 
hc = AgglomerativeClustering(n_clusters=2, linkage='ward')
summary_df['Hierarchical'] = hc.fit_predict(X_scaled)

# 3. DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)
summary_df['DBSCAN'] = dbscan.fit_predict(X_scaled)

# 4. Mean Shift
bandwidth = estimate_bandwidth(X_scaled, quantile=0.2, n_samples=500)
meanshift = MeanShift(bandwidth=bandwidth, bin_seeding=True)
summary_df['MeanShift'] = meanshift.fit_predict(X_scaled)

# 5. Gaussian Mixture Model (GMM)
gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)
summary_df['GMM'] = gmm.fit_predict(X_scaled)


print("\n=== STEP 4: MODELS EVALUATION SUMMARY ===")
models_list = ['KMeans', 'Hierarchical', 'DBSCAN', 'MeanShift', 'GMM']
metrics_summary = []

for m in models_list:
    labels = summary_df[m]
    unique_labels = len(np.unique(labels))
    
    # Calculate silhouette score only if clusters are valid (> 1)
    if 1 < unique_labels < len(X_scaled):
        sil = silhouette_score(X_scaled, labels)
        sil_str = f"{sil:.4f}"
    else:
        sil_str = "N/A (Density Noise/Single Cluster)"
        
    metrics_summary.append({
        'Model Name': m,
        'Clusters Found': unique_labels,
        'Silhouette Score': sil_str
    })

print(pd.DataFrame(metrics_summary).to_string(index=False))


print("\n--- Step 5: Generating Comparison Plots Matrix ---")
fig, axes = plt.subplots(5, 2, figsize=(15, 22))
plt.subplots_adjust(hspace=0.4)

for i, model_name in enumerate(models_list):
    # Left Column: PCA Space
    sns.scatterplot(
        x='PC1', y='PC2', hue=model_name, data=summary_df, 
        palette='viridis', ax=axes[i, 0], s=30, alpha=0.7
    )
    axes[i, 0].set_title(f'{model_name} Clustering (PCA View)')
    
    # Right Column: t-SNE Space
    sns.scatterplot(
        x='tSNE1', y='tSNE2', hue=model_name, data=summary_df, 
        palette='plasma', ax=axes[i, 1], s=30, alpha=0.7
    )
    axes[i, 1].set_title(f'{model_name} Clustering (t-SNE View)')

plt.suptitle('Comparison Matrix of All Unsupervised Learning Models', fontsize=16, y=0.92)
plt.show()


print("--- Step 6: Generating Hierarchical Tree Structure (Dendrogram) ---")
plt.figure(figsize=(10, 5))
sch.dendrogram(
    sch.linkage(X_scaled, method='ward'),
    truncate_mode='level', 
    p=4,                  
    color_threshold=25,   
    leaf_rotation=45.,
    leaf_font_size=10.
)
plt.title('Hierarchical Clustering Tree Structure (Dendrogram)')
plt.xlabel('Cluster Sizes / Node Counts')
plt.ylabel('Distance Threshold')
plt.axhline(y=25, color='r', linestyle='--', label='Tree Cut-off Threshold')
plt.legend()
plt.show()


print("--- Step 7: Final Step - Consensus Ensemble & Data Export ---")
# Finding voting consensus among the most stable models
top_models = ['KMeans', 'Hierarchical', 'GMM']
summary_df['Consensus_Cluster'] = summary_df[top_models].mode(axis=1)[0].astype(int)

# Attach final tags to original dataframe
final_df = df.copy()
final_df['Predicted_Group'] = summary_df['Consensus_Cluster']

print("\n========================================================")
print("             FINAL PATIENT PROFILING REPORT             ")
print("========================================================")
columns_to_show = ['Glucose', 'BMI', 'Age', 'BloodPressure', 'Insulin', 'Outcome']
print(final_df.groupby('Predicted_Group')[columns_to_show].mean().round(2).to_string())
print("========================================================\n")

# Exporting final results
output_file = 'final_diabetic_clusters_report.csv'
final_df.to_csv(output_file, index=False)

print(f"🎉 PROJECT COMPLETED SUCCESSFULLY!")
print(f"📁 Labeled dataset saved to your folder as: '{output_file}'")
```
